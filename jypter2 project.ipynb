```python
import os
import shutil

path = r"C:\Users\ADMIN\OneDrive\Attachments\Python Tutorials"
folder_names = ['csv files', 'image files', 'text files', 'excel files', 'word files']

for folder in folder_names:
    folder_path = os.path.join(path, folder)
    if not os.path.exists(folder_path):
        os.makedirs(folder_path)

file_names = os.listdir(path)

for file in file_names:
    source_file = os.path.join(path, file)
    
    if os.path.isfile(source_file):
        if ".csv" in file.lower():
            destination = os.path.join(path, "csv files", file)
            shutil.move(source_file, destination)
            
        elif ".png" in file.lower() or ".jpg" in file.lower():
            destination = os.path.join(path, "image files", file)
            shutil.move(source_file, destination)
            
        elif ".txt" in file.lower():
            destination = os.path.join(path, "text files", file)
            shutil.move(source_file, destination)
            
        elif ".xlsx" in file.lower() or ".xls" in file.lower():
            destination = os.path.join(path, "excel files", file)
            shutil.move(source_file, destination)
            
        elif ".docx" in file.lower() or ".doc" in file.lower():
            destination = os.path.join(path, "word files", file)
            shutil.move(source_file, destination)
```
